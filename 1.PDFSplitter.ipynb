# PDF Loader & Splitter (Hugging Face)

Here we loads PDFs, splits them into manageable chunks, and saves them for later use in RAG pipeline.

Steps:

Load PDFs from ./data/

Split into overlapping chunks using RecursiveCharacterTextSplitter

Inspect chunk statistics

Save to outputs/chunks.jsonl

```python
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
import json

# Ensure output directory exists
os.makedirs("outputs", exist_ok=True)
```

```python
# Path to data folder
data_path = "C:/Users/ADMIN/Documents/rag_chatbot/pdfs"
# Collect PDFs
pdf_files = [os.path.join(data_path, f) for f in os.listdir(data_path) if f.endswith(".pdf")]
print("Found PDFs:", pdf_files)

# Load all pages from all PDFs
docs = []
for pdf in pdf_files:
    loader = PyPDFLoader(pdf)
    pages = loader.load()
    docs.extend(pages)

print(f"Loaded {len(docs)} pages from {len(pdf_files)} PDFs")
```

```python
# Split documents into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100,
    length_function=len
)
chunks = splitter.split_documents(docs)
print(f"Created {len(chunks)} chunks")
```

```python
# Inspect a few chunks
for i, chunk in enumerate(chunks[:3]):
    print(f"--- Chunk {i+1} ---")
    print(chunk.page_content[:500])
    print()
```

```python
# Save chunks to outputs/chunks.jsonl
output_path = r"C:\Users\ADMIN\Documents\rag_chatbot\chunks.jsonl"
with open(output_path, "w", encoding="utf-8") as f:
    for chunk in chunks:
        record = {
            "text": chunk.page_content,
            "metadata": chunk.metadata
        }
        f.write(json.dumps(record) + "\n")

print(f"Saved chunks to {output_path}")
```
